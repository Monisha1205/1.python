```python
import pandas as pd

quadrant_results = {}

for year, data in datasets.items():

    country_col = country_columns[year]
    freedom_col = freedom_columns[year]
    score_col = score_columns[year]

    # Create a new DataFrame
    df = data[[country_col, freedom_col, score_col]].copy()

    # Find median values
    freedom_median = df[freedom_col].median()
    happiness_median = df[score_col].median()
    # helps us categorize countries into High and Low groups for the four combinations
    # Categorize Freedom
    df["Freedom_Level"] = df[freedom_col].apply(
        lambda x: "High" if x >= freedom_median else "Low"
    )

    # Categorize Happiness
    df["Happiness_Level"] = df[score_col].apply(
        lambda x: "High" if x >= happiness_median else "Low"
    )

    # Combine the categories
    df["Category"] = (
        df["Freedom_Level"] + " Freedom / " +
        df["Happiness_Level"] + " Happiness"
    )

    quadrant_results[year] = df

    print("\n" + "=" * 60)
    print(f"YEAR: {year}")
    print("=" * 60)

    print(df["Category"].value_counts()) # total country flows on the each category
```

# Q8. Compare happiness scores across years and identify countries where happiness increased, decreased, or remained stable.

```python
import pandas as pd

# Country column names
country_columns = {
    2015: "Country",
    2016: "Country",
    2017: "Country",
    2018: "Country",
    2019: "Country"
}


# Happiness Score column names
score_columns = {
    2015: "Happiness Score",
    2016: "Happiness Score",
    2017: "Happiness Score",
    2018: "Happiness Score",
    2019: "Happiness Score"
}


# Store yearly data
happiness_data = []


# Extract required columns
for year, data in datasets.items():

    country_col = country_columns[year]
    score_col = score_columns[year]

    temp = data[[country_col, score_col]].copy()

    # Standardize column names
    temp.columns = ["Country", "Happiness Score"]

    # Add Year
    temp["Year"] = year

    happiness_data.append(temp)


# Combine all datasets
combined_happiness = pd.concat(
    happiness_data,
    ignore_index=True
)


# Create country vs year table
happiness_pivot = combined_happiness.pivot(
    index="Country",
    columns="Year",
    values="Happiness Score"
)


# Calculate change from first available year to last available year
def calculate_change(row):

    values = row.dropna()

    if len(values) < 2:
        return None

    return values.iloc[-1] - values.iloc[0]


happiness_pivot["Change"] = happiness_pivot.apply(
    calculate_change,
    axis=1
)


# Classify change
def classify_change(change):

    if pd.isna(change):
        return "Insufficient Data"

    elif change > 0.05:
        return "Increased"

    elif change < -0.05:
        return "Decreased"

    else:
        return "Stable"


happiness_pivot["Status"] = happiness_pivot["Change"].apply(
    classify_change
)


# Results
print("\nCOUNTRIES WITH INCREASED HAPPINESS")
print(
    happiness_pivot[
        happiness_pivot["Status"] == "Increased"
    ]
)


print("\nCOUNTRIES WITH DECREASED HAPPINESS")
print(
    happiness_pivot[
        happiness_pivot["Status"] == "Decreased"
    ]
)


print("\nCOUNTRIES WITH STABLE HAPPINESS")
print(
    happiness_pivot[
        happiness_pivot["Status"] == "Stable"
    ]
)


# Summary
print("\nSUMMARY")
print(happiness_pivot["Status"].value_counts())
```

# Q9. Determine which factors have the strongest relationship with happiness: GDP/Economy, Social Support, Life Expectancy, Freedom, Generosity, and Corruption Perception.

```python
import pandas as pd

factors = [
    "GDP",
    "Social Support",
    "Life Expectancy",
    "Freedom",
    "Generosity",
    "Corruption Perception"
]

all_correlations = {}

for year, data in datasets.items():

    correlations = data[
        factors + ["Happiness Score"]
    ].corr()["Happiness Score"]

    all_correlations[year] = correlations[factors]


# Create correlation DataFrame
correlation_df = pd.DataFrame(all_correlations).T

print("YEAR-WISE CORRELATION")
print(correlation_df)


# Average correlation
average_correlation = correlation_df.mean()


# Create ranking
factor_ranking = pd.DataFrame({
    "Factor": average_correlation.index,
    "Average Correlation": average_correlation.values
})


# Absolute correlation
factor_ranking["Absolute Correlation"] = (
    factor_ranking["Average Correlation"].abs()
)


# Sort from strongest to weakest
factor_ranking = factor_ranking.sort_values(
    by="Absolute Correlation",
    ascending=False
).reset_index(drop=True)

factor_ranking
```

```python
plt.figure(figsize=(10, 6))

plt.barh(
    factor_ranking["Factor"],
    factor_ranking["Average Correlation"]
)

plt.xlabel("Average Correlation with Happiness Score")
plt.ylabel("Factors")

plt.title("Factors Related to Happiness Score (2015–2019)")

plt.gca().invert_yaxis()

plt.show()
```
