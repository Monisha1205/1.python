```python
dataset.drop(columns = ["sl_no"], inplace=True)
dataset.head()
```

```python
def quanqual(dataset):
  quan = []
  qual = []
  for ch in  dataset.columns:
    if dataset[ch].dtype == "O":
      #print("qual")
      qual.append(ch)
    else:
      #print("quan")
      quan.append(ch)
  return quan,qual
```

```python
quan,qual = quanqual(dataset)
```

Ascending Order Works Best

*   It shows the natural flow of numbers.
*   It makes it easy to spot the lowest and highest scores.
*   It keeps cumulative frequency correct.
*   dont arrange the marks according to frequency descending when calculating cumulative frequency. That would change the meaning of cumulative frequency.






```python
for ColumnsName in quan:
  Freq_table = pd.DataFrame(columns=("unique_values", "frequency", "relative_frequency", "cumsum"))
  Freq_table["unique_values"] = dataset[ColumnsName].value_counts().sort_index().index
  Freq_table["frequency"] = dataset[ColumnsName].value_counts().sort_index().values
  Freq_table["relative_frequency"] = Freq_table["frequency"]/dataset.shape[0]
  Freq_table["cumsum"] = round((Freq_table["relative_frequency"].cumsum())*100,2)
  print("\nFrequency Table for:", ColumnsName)
  print(Freq_table)
```
