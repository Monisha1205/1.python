```python
def quanqual(dataset):
  quan = []
  qual = []
  for ch in  dataset.columns:
    if dataset[ch].dtype == "O":
      #print("qual")
      qual.append(ch)
    else:
      #print("quan")
      quan.append(ch)
  return quan,qual
```

```python
quan,qual = quanqual(dataset)
```

```python
dataset1 =  dataset1.drop(columns = ["sl_no"])
```

```python
dataset1["ssc_p"].skew()
```

```python
skewness_kurtosis = pd.DataFrame(
    index=("skewness", "skew_type", "kurtosis", "kur_type"),
    columns=quan
)

for ColumnsName in quan:
    skewness_kurtosis[ColumnsName]["skewness"] = dataset1[ColumnsName].skew()
    skewness_kurtosis[ColumnsName]["skew_type"] = np.where(skewness_kurtosis[ColumnsName]["skewness"] < 0, "Negative skewness", np.where( skewness_kurtosis[ColumnsName]["skewness"] == 0, "Symmetric","Positive skewness") )
    skewness_kurtosis[ColumnsName]["kurtosis"] = dataset1[ColumnsName].kurtosis() + 3
    skewness_kurtosis[ColumnsName]["kur_type"] = np.where(skewness_kurtosis[ColumnsName]["kurtosis"] < 3,"Platykurtic",np.where(skewness_kurtosis[ColumnsName]["kurtosis"] == 3,"Mesokurtic","Leptokurtic"))
```
